```python
!pip -q install arabic-reshaper python-bidi

import matplotlib.pyplot as plt
import arabic_reshaper
from bidi.algorithm import get_display

def arabic_text(text):
    return get_display(arabic_reshaper.reshape(str(text)))

chart_data = df.dropna(subset=["نسبة_الإنجاز"]).copy()

plt.figure(figsize=(14, 7))

labels = [arabic_text(x) for x in chart_data["المؤشر"]]

plt.bar(labels, chart_data["نسبة_الإنجاز"])

plt.title(arabic_text("نسبة إنجاز مؤشرات رؤية السعودية 2030"))
plt.xlabel(arabic_text("المؤشرات"))
plt.ylabel("Achievement (%)")

plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

chart_data = df.dropna(subset=["المستهدف_2030"])

labels = [arabic_text(x) for x in chart_data["المؤشر"]]
x = np.arange(len(labels))
width = 0.35

plt.figure(figsize=(14, 7))

plt.bar(x - width/2, chart_data["القيمة"], width, label="القيمة الحالية")
plt.bar(x + width/2, chart_data["المستهدف_2030"], width, label="مستهدف 2030")

plt.title(arabic_text("مقارنة القيمة الحالية بالمستهدف لعام 2030"))
plt.xlabel(arabic_text("المؤشرات"))
plt.ylabel(arabic_text("القيمة"))
plt.xticks(x, labels, rotation=45, ha="right")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
axis_data = df.groupby("المحور")["نسبة_الإنجاز"].mean().dropna()

labels = [arabic_text(x) for x in axis_data.index]

plt.figure(figsize=(9, 6))
plt.bar(labels, axis_data.values)

plt.title(arabic_text("متوسط نسبة الإنجاز حسب المحور"))
plt.xlabel(arabic_text("المحور"))
plt.ylabel(arabic_text("متوسط نسبة الإنجاز (%)"))

plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")

plot_data = df.dropna(subset=["القيمة", "المستهدف_2030"]).copy()

plt.figure(figsize=(10, 7))

sns.scatterplot(
    data=plot_data,
    x="المستهدف_2030",
    y="القيمة",
    hue="المحور",
    s=120
)
plt.title(arabic_text("العلاقة بين القيمة الحالية والمستهدف"))
plt.xlabel(arabic_text("المستهدف لعام 2030"))
plt.ylabel(arabic_text("القيمة الحالية"))
legend = plt.legend()

for item in legend.get_texts():
    item.set_text(arabic_text(item.get_text()))

legend.set_title(arabic_text("المحور"))
plt.tight_layout()
plt.show()
```
